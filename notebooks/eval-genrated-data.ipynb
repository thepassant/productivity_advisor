```python
import pandas as pd 
import json 
from tqdm.auto import tqdm
```

```python
from openai import OpenAI

client = OpenAI()
```

```python
prompt1_template = """
You are a productivity expert generating evaluation questions.
For the productivity task below, generate 2 questions that a user might ask.
Return the result as a JSON array with objects containing 'id' and 'question' fields.

Task: {task}
Category: {category}
Difficulty: {difficulty}
Duration: {duration_estimate}
Instructions: {instructions}
Reasoning: {reasoning}
Tags: {tags}
""".strip()
```

```python
df = pd.read_csv('../data/cleaned_data.csv')
documents = df.to_dict(orient='records')
```

```python
results = []

for idx, row in tqdm(df.iterrows(), total=len(df)):
    prompt = prompt1_template.format(**row.to_dict())

    response = client.responses.create(
        model="gpt-5.4-mini",
        input=[{"role": "user", "content": prompt}]
    )

    questions = json.loads(response.output_text)

    for q in questions:
        q["task_id"] = idx
        results.append(q)

df_questions = pd.DataFrame(results)
df_questions.to_csv("../data/ground-truth-retrieval.csv", index=False)
```
